```julia
using VeryDiff, LinearAlgebra, Plots, Measures
import VeryDiff: approximate_polynomial_iterative
```

```julia
W1 = [1 1; 1 -1.]
b1 = zeros(2)
W2 = [1 1; 1 -1.]
b2 = [-0.5, 0]
W3 = [-1 0; 1 1.]
b3 = zeros(2)

W1a = [1 1.1; 1 -0.9]
b1a = zeros(2)
W2a = [1.1 1; 0.9 -1]
b2a = [-0.4, 0]
W3a = [-1 0; 1 1.1]
b3a = zeros(2)

layers = [Dense(W1, b1), ReLU(), Dense(W2, b2), ReLU(), Dense(W3, b3)]
layersa = [Dense(W1a, b1a), ReLU(), Dense(W2a, b2a), ReLU(), Dense(W3a, b3a)];
```

```julia
bounds_diff = zono_bounds(ẑΔ.∂Z)
```

```julia
function verify_epsilon_equivalence(nn1, nn2, z)
    nn_diff = GeminiNetwork(nn1, nn2)

    ∂z = Zonotope(zero(z.G), zero(z.c), nothing)
    zΔ = DiffZonotope(z, deepcopy(z), ∂z, 0, 0, 0)

    ẑΔ = nn_diff(zΔ, PropState(true))
    return ẑΔ
end
```

# After 1st Linear Layer

```julia
idx = 1
nn = Network(layers[1:idx])
nna = Network(layersa[1:idx])

z = Zonotope(-ones(2), ones(2))

zΔ = verify_epsilon_equivalence(nn, nna, z)

println("Z₁: c = $(zΔ.Z₁.c), G = $(zΔ.Z₁.G)")
println("Z₂: c = $(zΔ.Z₂.c), G = $(zΔ.Z₂.G)")
println("∂Z: c = $(zΔ.∂Z.c), G = $(zΔ.∂Z.G)")
```
