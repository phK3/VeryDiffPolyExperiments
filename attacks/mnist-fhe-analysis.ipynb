```python
import re

fhe_label_parsed = []

# Split by image sections
image_sections = fhe_results.split("=== img_")

for section in image_sections[1:]:  # Skip the first empty split
    lines = section.split("\n")
    
    # Check if there's an error in this section
    if "terminating due to uncaught exception" in section or "libc++abi" in section:
        fhe_label_parsed.append(-1)
    else:
        # Look for "Output of the network (encrypted):" line
        result_found = False
        for line in lines:
            if "Output of the network (encrypted):" in line:
                # Extract the number after "Output of the network (encrypted):"
                match = re.search(r"Output of the network \(encrypted\):\s*(\[.*?\])", line)
                if match:
                    # Parse list of numbers
                    # example: [ -2.048156537 -0.8572904147 -2.083540534 -1.918192692 1.262908742 -3.727712507 -1.545996529 2.964452447 -0.8939218776 1.607958483 ]
                    numbers_str = match.group(1)
                    numbers = [float(num) for num in re.findall(r"[-+]?\d*\.\d+|[-+]?\d+", numbers_str)]
                    # The predicted label is the argmax of the numbers
                    predicted_label = numbers.index(max(numbers))
                    fhe_label_parsed.append(predicted_label)
                    result_found = True
                    break

        if not result_found:
            fhe_label_parsed.append(-1)
print(f"Parsed {len(fhe_label_parsed)} FHE results")
print(f"First 20 labels: {fhe_label_parsed[:20]}")
```
