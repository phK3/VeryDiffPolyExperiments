## Plots

### Approximation Error

```julia
yticks = [1,2,5,10,20,50,100, 200, 500, 1000, 2000, 5000]
ytick_labels = string.(yticks)
plot(xlabel="degree", ylabel="error", yaxis=:log10, yticks=(yticks, ytick_labels), formatter=:plain, legend=:topright)

plot!(collins_gelu_single_results["degrees"], collins_gelu_single_results["ϵs"], label="GELU single (verified)", markershape=:circle, color=1)
plot!(collins_gelu_single_results["degrees"], collins_gelu_single_results["ϵ_samples"], label="GELU single (sampled)", linestyle=:dash, markershape=:circle, color=1)

plot!(collins_gelu_results["degrees"], collins_gelu_results["ϵs"], label="GELU (verified)", markershape=:circle, color=2)
plot!(collins_gelu_results["degrees"], collins_gelu_results["ϵ_samples"], label="GELU (sampled)", linestyle=:dash, markershape=:circle, color=2)

plot!(collins_gelu_full_results["degrees"], collins_gelu_full_results["ϵs"], label="GELU full (verified)", markershape=:circle, color=3)
plot!(collins_gelu_full_results["degrees"], collins_gelu_full_results["ϵ_samples"], label="GELU full (sampled)", linestyle=:dash, markershape=:circle, color=3)
```

### Approximation Time

```julia
plot(xlabel="degree", ylabel="approximation time", legend=:topleft)

plot!(collins_gelu_single_results["degrees"], collins_gelu_single_results["t_approxs"], label="GELU single", markershape=:circle, color=1)
plot!(collins_gelu_results["degrees"], collins_gelu_results["t_approxs"], label="GELU", markershape=:circle, color=2)
plot!(collins_gelu_full_results["degrees"], collins_gelu_full_results["t_approxs"], label="GELU full", markershape=:circle, color=3)
```

### Verification Time

```julia
plot(xlabel="degree", ylabel="approximation time", legend=:topleft)

plot!(collins_gelu_single_results["degrees"], collins_gelu_single_results["t_verifies"], label="GELU single", markershape=:circle, color=1)
plot!(collins_gelu_results["degrees"], collins_gelu_results["t_verifies"], label="GELU", markershape=:circle, color=2)
plot!(collins_gelu_full_results["degrees"], collins_gelu_full_results["t_verifies"], label="GELU full", markershape=:circle, color=3)
```
