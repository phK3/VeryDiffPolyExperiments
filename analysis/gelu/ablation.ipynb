## Plot

```julia
plot(xlabel="degree", ylabel="error", yaxis=:log10, yticks=[1e-3, 1e-2, 1e-1, 1, 1e1, 1e2, 1e3], ylims=(1e-3, 2000), legend=:bottomleft)

plot!(mnist_relu_single_results["degrees"], mnist_relu_single_results["ϵs"], label="ReLU single (verified)", linestyle=:dot, markershape=:circle, color=1)
plot!(mnist_gelu_single_results["degrees"], mnist_gelu_single_results["ϵs"], label="GELU single(verified)", linestyle=:dot, markershape=:circle, color=2)

plot!(mnist_relu_single_results["degrees"], mnist_relu_single_results["ϵ_samples"], label="ReLU single (sampled)", linestyle=:dashdot, markershape=:circle, color=1)
plot!(mnist_gelu_single_results["degrees"], mnist_gelu_single_results["ϵ_samples"], label="GELU single(sampled)", linestyle=:dashdot, markershape=:circle, color=2)

plot!(mnist_relu_results["degrees"], mnist_relu_results["ϵs"], label="ReLU (verified)", markershape=:circle, color=1)
plot!(mnist_gelu_results["degrees"], mnist_gelu_results["ϵs"], label="GELU (verified)", markershape=:circle, color=2)

plot!(mnist_relu_results["degrees"], mnist_relu_results["ϵ_samples"], label="ReLU (sampled)", markershape=:circle, linestyle=:dash, color=1)
plot!(mnist_gelu_results["degrees"], mnist_gelu_results["ϵ_samples"], label="GELU (sampled)", markershape=:circle, linestyle=:dash, color=2)
```
