### Number Values for Improvement

```julia
function compare_config(results1, results2; sampled=false, sampled1=false, sampled2=false)
    if sampled 
        sampled1 = true
        sampled2 = true
    end
    if sampled1
        ϵ1 = results1["ϵ_samples"]
    else 
        ϵ1 = results1["ϵs"]
    end

    if sampled2
        ϵ2 = results2["ϵ_samples"]
    else
        ϵ2 = results2["ϵs"]
    end
    
    println("config1 ϵs: ", ϵ1)
    println("config2 ϵs: ", ϵ2)
    println("ratio:      ", ϵ1 ./ ϵ2)
end
```

```julia
println("# ReLU vs GELU")
println("--- mnist ---")
println("\nsingle")
compare_config(mnist_relu_single_results, mnist_gelu_single_results)
println("\nmulti")
compare_config(mnist_relu_results, mnist_gelu_results)
println("\nsingle sampling")
compare_config(mnist_relu_single_sampling_results, mnist_gelu_single_sampling_results, sampled=true)
println("\nmulti sampling")
compare_config(mnist_relu_sampling_results, mnist_gelu_sampling_results, sampled=true)

println("\n--- collins ---")
println("\nsingle")
compare_config(collins_relu_single_results, collins_gelu_single_results)
println("\nmulti")
compare_config(collins_relu_results, collins_gelu_results)
println("\nsingle sampling")
compare_config(collins_relu_single_sampling_results, collins_gelu_single_sampling_results, sampled=true)
println("\nmulti sampling")
compare_config(collins_relu_sampling_results, collins_gelu_sampling_results, sampled=true)

println("\n\n# Uniform vs Heterogenuous")
println("--- mnist ---")
println("\nrelu")
compare_config(mnist_relu_single_results, mnist_relu_results)
println("\ngelu")
compare_config(mnist_gelu_single_results, mnist_gelu_results)
println("\nrelu sampling")
compare_config(mnist_relu_single_sampling_results, mnist_relu_sampling_results, sampled=true)
println("\ngelu sampling")
compare_config(mnist_gelu_single_sampling_results, mnist_gelu_sampling_results, sampled=true)

println("\n--- collins ---")
println("\nrelu")
compare_config(collins_relu_single_results, collins_relu_results)
println("\ngelu")
compare_config(collins_gelu_single_results, collins_gelu_results)
println("\nrelu sampling")
compare_config(collins_relu_single_sampling_results, collins_relu_sampling_results, sampled=true)
println("\ngelu sampling")
compare_config(collins_gelu_single_sampling_results, collins_gelu_sampling_results, sampled=true)

println("\n\n# Verified vs Sampling")
println("--- mnist ---")
println("\nrelu single")
compare_config(mnist_relu_single_results, mnist_relu_single_sampling_results, sampled2=true)
println("\ngelu single")
compare_config(mnist_gelu_single_results, mnist_gelu_single_sampling_results, sampled2=true)
println("\nrelu multi")
compare_config(mnist_relu_results, mnist_relu_sampling_results, sampled2=true)
println("\ngelu multi")
compare_config(mnist_gelu_results, mnist_gelu_sampling_results, sampled2=true)

println("\n--- collins ---")
println("\nrelu single")
compare_config(collins_relu_single_results, collins_relu_single_sampling_results, sampled2=true)
println("\ngelu single")
compare_config(collins_gelu_single_results, collins_gelu_single_sampling_results, sampled2=true)
println("\nrelu multi")
compare_config(collins_relu_results, collins_relu_sampling_results, sampled2=true)
println("\ngelu sampling")
compare_config(collins_gelu_results, collins_gelu_sampling_results, sampled2=true)
```

### Analysis of Pre-Activation Ranges

```julia
get_bounds_range(node::VeryDiff.ONNXChebyshevPoly) = (minimum(node.l), maximum(node.u))
get_bounds_range(node) = nothing

function get_bounds_range(net::VNNLib.OnnxParser.OnnxNet)
    for (k, v) in net.nodes
        r = get_bounds_range(v)
        if !isnothing(r)
            println(k, ": [", first(r), ", ", last(r), "]")
        end
    end
end
```

For MNIST, the GELU networks converge a lot faster than for Collins, because the pre-activation bounds are a lot smaller - already for the sampled bounds.
This is in line with our expectations since for large approximation intervals, GELU and ReLU behave similarly.
