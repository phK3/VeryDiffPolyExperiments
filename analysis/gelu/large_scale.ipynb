```julia
get_l1_norm(L::VNNLib.OnnxParser.ONNXLinear) = sum(abs.(L.dense.weight))
get_l1_norm(L::VNNLib.OnnxParser.ONNXConv) = sum(abs.(L.conv.weight))
get_l1_norm(L) = 0.  # default case for layers without weights

get_l1_norm(model::VNNLib.OnnxParser.OnnxNet) = sum(get_l1_norm(layer) for (layer_name, layer) in model.nodes)
```
